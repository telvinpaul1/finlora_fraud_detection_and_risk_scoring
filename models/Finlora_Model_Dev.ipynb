# Finlora Finance Flow — Fraud Detection & Risk Scoring
## Model Development

This notebook develops two fraud-classification models:
1. Logistic Regression — interpretable baseline.
2. Random Forest — non-linear prototype.

The pipeline uses a stratified 75/25 train-test split and SMOTE on the training data only.

```python
#  IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from imblearn.over_sampling import SMOTE

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score, precision_score, precision_recall_curve, recall_score, f1_score,
    roc_auc_score, confusion_matrix, ConfusionMatrixDisplay,
    classification_report
)
import joblib
import os
import pickle

print("Libraries imported successfully.")
```

##  Load the Feature-Engineered Dataset

The dataset already contains the original variables and engineered features such as log transformations, cyclical time features, behavioural flags and account/device features.

```python
#  LOAD DATASET

file_path = r"C:\Users\telvi\Downloads\AMDARI\finlora_finance_flow\data\processed_data\finlora_features.csv"

df = pd.read_csv(file_path)
```

```python
print("Dataset shape:", df.shape)

print("\nFraud distribution:")
print(df["is_fraud"].value_counts())

print("\nFraud percentage:")
print(df["is_fraud"].value_counts(normalize=True).mul(100).round(2))
```

##  Inspect Engineered Features

```python
#  ENGINEERED FEATURES

engineered_features = [
    "log_amount", "log_amount_to_avg_ratio",
    "hour_sin", "hour_cos", "day_sin", "day_cos",
    "high_amount_ratio_flag", "high_velocity_flag",
    "new_device_flag", "cross_border_flag",
    "account_age_group", "device_status_unknown"
]

print("Engineered features:")
for feature in engineered_features:
    print(f"- {feature}")

display(df[engineered_features].head(10))
```

# MODEL DEVELOPMENT

## Define Predictors and Target

Identifiers, raw text, dates and the potentially post-outcome `status` field are excluded. The target `is_fraud` is kept separately.

```python
#  DEFINE FEATURES (X) AND TARGET (y)

drop_features = [
    "transaction_id",
    "account_id",
    "description",
    "merchant_name",
    "device_id",
    "account_holder_name",
    "timestamp",
    "account_created_date",
    "status",
    "is_fraud"
]

X = df.drop(columns=drop_features)
y = df["is_fraud"]

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

## Identify Categorical and Numerical Features

This step prevents the earlier `ValueError: could not convert string to float: 'Wednesday'`.

Text/object columns such as `day_of_week` are treated as categorical and will be one-hot encoded.

```python
#  IDENTIFY CATEGORICAL AND NUMERICAL FEATURES

categorical_features = X.select_dtypes(include=["object"]).columns.tolist()
numeric_features = X.select_dtypes(exclude=["object"]).columns.tolist()

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numeric_features)
```

## Stratified 75/25 Train-Test Split

Stratification preserves the fraud/legitimate class proportions in both sets.

```python
# STRATIFIED TRAIN-TEST SPLIT

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    stratify=y,
    random_state=42
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTest target distribution:")
print(y_test.value_counts())
```

##  One-Hot Encode Categorical Features

The encoder learns from the training data only. `handle_unknown="ignore"` prevents errors if a category appears in the test set but was not present in training.

```python
#  ONE-HOT ENCODING

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore", sparse_output=False),
            categorical_features
        )
    ],
    remainder="passthrough"
)

X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)

print("Encoded training shape:", X_train_encoded.shape)
print("Encoded test shape:", X_test_encoded.shape)
```

##  Handle Class Imbalance with SMOTE

SMOTE is applied **only to the training data**. The test data remains untouched so that final evaluation uses the original class distribution.

```python
#  SMOTE — TRAINING DATA ONLY

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_encoded,
    y_train
)

print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(y_train_smote.value_counts())
```

# LOGISTIC REGRESSION

## Scale Data

Scaling is used for Logistic Regression. Random Forest does not require feature scaling.

```python
# CALE DATA FOR LOGISTIC REGRESSION

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_smote)
X_test_scaled = scaler.transform(X_test_encoded)

print("Scaling completed.")
```

```python
#  TRAIN LOGISTIC REGRESSION

# Logistic Regression is used as the interpretable baseline.
# SMOTE has already balanced the training data.
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train_scaled, y_train_smote)

print("Logistic Regression training completed.")
```

# RANDOM FOREST

## Train Random Forest

Random Forest can model non-linear relationships and interactions between fraud-related features.

```python
# TRAIN RANDOM FOREST

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_smote, y_train_smote)

print("Random Forest training completed.")
```

# MODEL PREDICTIONS

## 12. Logistic Regression Predictions

`y_pred_lr` contains binary predictions. `y_proba_lr` contains the estimated probability of fraud.

```python
# LOGISTIC REGRESSION PREDICTIONS

y_pred_lr = logistic_model.predict(X_test_scaled)
y_proba_lr = logistic_model.predict_proba(X_test_scaled)[:, 1]

print("Logistic Regression predictions generated.")
print("\nFirst 10 predictions:")
print(y_pred_lr[:10])

print("\nFirst 10 fraud probabilities:")
print(np.round(y_proba_lr[:10], 4))
```

##  Random Forest Predictions

```python
# RANDOM FOREST PREDICTIONS

y_pred_rf = rf_model.predict(X_test_encoded)
y_proba_rf = rf_model.predict_proba(X_test_encoded)[:, 1]

print("Random Forest predictions generated.")
print("\nFirst 10 predictions:")
print(y_pred_rf[:10])

print("\nFirst 10 fraud probabilities:")
print(np.round(y_proba_rf[:10], 4))
```

# MODEL EVALUATION

## Evaluation Metrics

Accuracy is reported for completeness, but precision, recall, F1-score and ROC-AUC are more informative for this imbalanced fraud problem.

```python
# MODEL EVALUATION FUNCTION

def evaluate_model(model_name, y_true, y_pred, y_proba):
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)
    roc_auc = roc_auc_score(y_true, y_proba)

    print(f"\n{model_name}")
    print("=" * 50)
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1-score : {f1:.4f}")
    print(f"ROC-AUC  : {roc_auc:.4f}")

    return {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "ROC-AUC": roc_auc
    }
```

```python
# EVALUATE LOGISTIC REGRESSION

lr_metrics = evaluate_model(
    "Logistic Regression",
    y_test,
    y_pred_lr,
    y_proba_lr
)
```

```python
# EVALUATE RANDOM FOREST

rf_metrics = evaluate_model(
    "Random Forest",
    y_test,
    y_pred_rf,
    y_proba_rf
)
```

## Compare Model Performance

```python
# MODEL PERFORMANCE COMPARISON

results = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "Accuracy": [lr_metrics["Accuracy"], rf_metrics["Accuracy"]],
    "Precision": [lr_metrics["Precision"], rf_metrics["Precision"]],
    "Recall": [lr_metrics["Recall"], rf_metrics["Recall"]],
    "F1-Score": [lr_metrics["F1-Score"], rf_metrics["F1-Score"]],
    "ROC-AUC": [lr_metrics["ROC-AUC"], rf_metrics["ROC-AUC"]]
})

display(results.round(4))
```

# FIXED-RECALL OPERATING POINT ANALYSIS

## Precision at 70%, 80% and 90% Recall

The default classification threshold of 0.50 is not necessarily the most appropriate operating point for fraud detection.

To support different fraud-review capacities, model thresholds are evaluated at target recall levels of 70%, 80% and 90%.

For each target recall level, the analysis identifies a probability threshold that achieves at least the required recall and reports the resulting precision.

This allows the risk team to choose an operating point based on the desired balance between:

- detecting fraudulent transactions (recall), and
- limiting the number of legitimate transactions sent for review (precision).

The analysis is performed on the untouched test dataset.

```python

# FIXED-RECALL OPERATING POINT ANALYSIS

# Target recall levels required by the risk team
target_recalls = [0.70, 0.80, 0.90]


def fixed_recall_analysis(y_true, y_proba, model_name):
    """
    Find probability thresholds that achieve at least
    the requested recall levels.

    For each target recall, the function reports:
    - probability threshold
    - achieved recall
    - precision
    - percentage of transactions flagged for review
    """

    # Calculate precision, recall and thresholds
    precision, recall, thresholds = precision_recall_curve(
        y_true,
        y_proba
    )

    results = []

    for target_recall in target_recalls:

        # Find thresholds where recall is at least
        # the required target.
        valid_indices = np.where(
            recall[:-1] >= target_recall
        )[0]

        # If the target recall cannot be achieved,
        # record missing values.
        if len(valid_indices) == 0:

            results.append({
                "Model": model_name,
                "Target Recall": target_recall,
                "Threshold": np.nan,
                "Precision": np.nan,
                "Achieved Recall": np.nan,
                "Transactions Flagged (%)": np.nan
            })

            continue

        # Select the highest threshold that still achieves
        # the target recall.
        # A higher threshold generally means fewer transactions
        # are sent for review, while still meeting the recall
        # requirement.
        selected_index = valid_indices[-1]

        selected_threshold = thresholds[selected_index]

        selected_precision = precision[selected_index]

        selected_recall = recall[selected_index]

        # Predict transactions using the selected threshold
        selected_predictions = (
            y_proba >= selected_threshold
        ).astype(int)

        # Calculate percentage of transactions flagged
        # as potentially fraudulent.
        flagged_percentage = (
            selected_predictions.sum()
            / len(selected_predictions)
            * 100
        )

        results.append({
            "Model": model_name,
            "Target Recall": target_recall,
            "Threshold": selected_threshold,
            "Precision": selected_precision,
            "Achieved Recall": selected_recall,
            "Transactions Flagged (%)": flagged_percentage
        })

    return pd.DataFrame(results)
```

```python
# ============================================================
# CHECK MODEL PREDICTIONS
# ============================================================

print("Test observations:", len(y_test))

print("\nRandom Forest probabilities:")
print(len(y_proba_rf))

print("\nLogistic Regression probabilities:")
print(len(y_proba_lr))

print("\nFirst 10 Random Forest probabilities:")
print(y_proba_rf[:10])
```

```python
# fixed-recall analysis for Random Forest

# Calculate precision, recall and thresholds
precision, recall, thresholds = precision_recall_curve(
    y_test,
    y_proba_rf
)

target_recalls = [0.70, 0.80, 0.90]

results = []

for target in target_recalls:

    # Find thresholds where recall is at least the target
    valid_indices = np.where(recall[:-1] >= target)[0]

    if len(valid_indices) > 0:

        # Use the highest threshold that still achieves the target recall
        index = valid_indices[-1]

        threshold = thresholds[index]

        # Apply threshold
        y_pred_threshold = (y_proba_rf >= threshold).astype(int)

        # Calculate actual metrics
        tp = ((y_test == 1) & (y_pred_threshold == 1)).sum()
        fp = ((y_test == 0) & (y_pred_threshold == 1)).sum()
        fn = ((y_test == 1) & (y_pred_threshold == 0)).sum()

        precision_value = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall_value = tp / (tp + fn) if (tp + fn) > 0 else 0

        flagged_percentage = (
            y_pred_threshold.sum() / len(y_pred_threshold)
        ) * 100

        results.append({
            "Target Recall": target,
            "Threshold": threshold,
            "Precision": precision_value,
            "Achieved Recall": recall_value,
            "Transactions Flagged (%)": flagged_percentage
        })

# Create results table
rf_recall_results = pd.DataFrame(results)

display(rf_recall_results.round(4))
```

The Random Forest model was evaluated at fixed recall operating points of 70%, 80%, and 90%. At 70% recall, the model achieved 58.23% precision while flagging 3.20% of transactions. Increasing recall to 80% reduced precision to 37.72% and increased the review volume to 5.65%. At 90% recall, precision declined to 13.44%, while 17.84% of transactions were flagged. These results demonstrate the precision–recall trade-off and provide tunable thresholds that can be selected according to the risk team's review capacity and tolerance for missed fraud.

```python
# For  Logistic Regression
precision, recall, thresholds = precision_recall_curve(
    y_test,
    y_proba_lr
)

results = []

for target in [0.70, 0.80, 0.90]:

    valid_indices = np.where(recall[:-1] >= target)[0]

    if len(valid_indices) > 0:

        index = valid_indices[-1]

        threshold = thresholds[index]

        y_pred_threshold = (y_proba_lr >= threshold).astype(int)

        tp = ((y_test == 1) & (y_pred_threshold == 1)).sum()
        fp = ((y_test == 0) & (y_pred_threshold == 1)).sum()
        fn = ((y_test == 1) & (y_pred_threshold == 0)).sum()

        precision_value = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall_value = tp / (tp + fn) if (tp + fn) > 0 else 0

        flagged_percentage = (
            y_pred_threshold.sum() / len(y_pred_threshold)
        ) * 100

        results.append({
            "Target Recall": target,
            "Threshold": threshold,
            "Precision": precision_value,
            "Achieved Recall": recall_value,
            "Transactions Flagged (%)": flagged_percentage
        })

lr_recall_results = pd.DataFrame(results)

display(lr_recall_results.round(4))
```

Fixed-recall analysis demonstrated a clear precision–recall trade-off for the Random Forest model. At 70% recall, the model achieved 65.44% precision while flagging only 2.85% of transactions. Increasing the target recall to 80% reduced precision to 38.74% and increased the flagged proportion to 5.50%. At 90% recall, precision decreased to 14.79%, while 16.21% of transactions were flagged. These results provide the Finlora risk team with tunable probability thresholds that can be selected according to fraud-loss tolerance and available investigation capacity.
